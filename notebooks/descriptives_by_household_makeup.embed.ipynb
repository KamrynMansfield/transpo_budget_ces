```python
# the combined data we put in the model
exp_data_full <- read.csv("../data/diary24_combined/created_data/exp_fmly_data.csv")



nrow(exp_data_full)
```

I ended up taking this data, formatting it in excel, and using it from
the excel sheet.

```python
#| label: tbl-descriptives-by-hhtype
#| tbl-cap: "Demograhic distribution grouped by differenct household structures"
#| tbl-alt: "A table with the shares of different demographic variables in the data split up by household type"

profiled_data <- exp_data_full |>
  mutate(profile = case_when(
    married_kids == 1 ~ "Married with children",
    married_no_kids == 1 ~ "Married without children",
    single_kids == 1 ~ "Not married but with children",
    single_no_kids == 1 ~ "Not married and no children",
    TRUE ~ "Other"
  )) |>
  select(any_of(col_names_to_use$variable), profile)

new_data <- profiled_data |>
  group_by(profile) |>
  summarise(n = n(),across(any_of(col_names_to_use$variable),  \(x) mean(x, na.rm = TRUE)))

transposed_data <- t(new_data) |> as.data.frame()

names(transposed_data) <- transposed_data[1,] |> as.character()

transposed_data <- transposed_data[2:nrow(transposed_data),]

transposed_data[] <- lapply(transposed_data, as.numeric)

nice_data <- transposed_data |>
  mutate(variable = rownames(transposed_data)) |>
  left_join(col_names_to_use, by = "variable") |>
  mutate("Married with children" = paste0(round(`Married with children` *100,1),"%"),
         "Married without children" = paste0(round(`Married without children` *100,1),"%"),
         "Not married but with children" = paste0(round(`Not married but with children` *100,1),"%"),
         "Not married and no children" = paste0(round(`Not married and no children` *100,1),"%"))

row_vals <- nice_data[nice_data$variable == "n",] |> as.character()
nice_data[nice_data$variable == "n",] <- ifelse(grepl("%",row_vals),
                                                as.character(as.numeric(substr(row_vals, 1, nchar(row_vals)-1)) / 100),
                                                row_vals)
```

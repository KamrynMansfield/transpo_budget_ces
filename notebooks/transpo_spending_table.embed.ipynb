# Category Spending Tables

```python
library(tidyverse)
```

```python
#| label: tbl-transpo-spending
#| tbl-cap: "Average Transportation Spending per Household in Each Category"
#| tbl-alt: "A table of Transportation Average Spending per Household in Each Category"

transpo_exp |> 
  mutate(pct_zeros = avg_exp_zeros * 100 / sum(avg_exp_zeros),
         pct_no_zeros = avg_exp_no_zeros * 100 / sum(avg_exp_no_zeros)) |>
  arrange(desc(count)) |>
  mutate(avg_exp_zeros = paste0("$",round(avg_exp_zeros, 2), " (",round(pct_zeros,2),"%)"),
         avg_exp_no_zeros = paste0("$",round(avg_exp_no_zeros, 2), " (",round(pct_no_zeros,2),"%)"),
         count_pct = paste0(count, " (",round(count * 100 / 6004,1),"% of sample)")) |>
  
  select(`Spending Category` = category, 
         `Average spending including all households` = avg_exp_zeros,
         `Average spending for only households that spent some in this category` = avg_exp_no_zeros,
         `Households with spending in this category` = count_pct) |>
  kable()
```
